##  Step 2: Review the Contract

Our contract defines the expected IoT telemetry schema and quality rules.

````python
from pathlib import Path
from lakelogic import DataProcessor

BASE = Path.cwd()
contract_path = BASE / "pubsub_contract.yaml"
if not contract_path.exists():
    candidate = BASE / "examples" / "03_data_sources" / "streaming" / "gcp_pubsub" / "pubsub_contract.yaml"
    if candidate.exists():
        BASE = candidate.parent
        contract_path = candidate

RUN_LIVE = False  # Set True when your streaming infrastructure is available

sample_events = [{'deviceId': 'device-001', 'temperature': 22.5, 'humidity': 45.0, 'pressure': 101.2, 'timestamp': '2026-02-15T10:30:00', 'location': {'lat': 47.6, 'lon': -122.3}}, {'deviceId': 'device-002', 'temperature': 25.1, 'humidity': 39.0, 'pressure': 100.8, 'timestamp': '2026-02-15T10:31:00', 'location': {'lat': 37.7, 'lon': -122.4}}]

if RUN_LIVE:
    print("Live streaming is disabled by default in this demo.")
    print("Set RUN_LIVE = True and configure credentials/endpoints to stream.")
else:
    processor = DataProcessor(contract=contract_path)
    result = processor.run(sample_events, source_path="sample")
    print(result)
    print(f"Good: {len(result.good)} | Bad: {len(result.bad)}")
````
