##  Step 1: Setup Azure Service Bus

```bash

```

##  Step 2: Review the Contract

Our contract defines the expected Service Bus message schema and quality rules.

````python
from pathlib import Path
from lakelogic import DataProcessor

BASE = Path.cwd()
contract_path = BASE / "service_bus_contract.yaml"
if not contract_path.exists():
    candidate = BASE / "examples" / "03_data_sources" / "streaming" / "azure_service_bus" / "service_bus_contract.yaml"
    if candidate.exists():
        BASE = candidate.parent
        contract_path = candidate

RUN_LIVE = False  # Set True when your streaming infrastructure is available

sample_events = [{'notificationId': 'N-001', 'userId': 'U-101', 'type': 'email', 'message': 'Welcome!', 'priority': 'high', 'timestamp': '2026-02-15T10:20:00', 'metadata': {'campaign': 'onboarding'}}, {'notificationId': 'N-002', 'userId': 'U-102', 'type': 'sms', 'message': 'Your code is 1234', 'priority': 'medium', 'timestamp': '2026-02-15T10:21:00', 'metadata': {'region': 'US'}}]

if RUN_LIVE:
    print("Live streaming is disabled by default in this demo.")
    print("Set RUN_LIVE = True and configure credentials/endpoints to stream.")
else:
    processor = DataProcessor(contract=contract_path)
    result = processor.run(sample_events, source_path="sample")
    print(result)
    print(f"Good: {len(result.good)} | Bad: {len(result.bad)}")
````
