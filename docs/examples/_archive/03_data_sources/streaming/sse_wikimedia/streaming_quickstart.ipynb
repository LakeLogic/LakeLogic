## Business Scenario

Live change streams like Wikimedia edits are a fast way to understand streaming ingestion. You need a simple path to validate and store events.

## Value Proposition

- Quickly validate live events
- Use a single contract for schema and rules
- Gain a repeatable streaming pattern

---

## Goals

1. Connect to an SSE stream
2. Validate live events
3. Inspect output


##  Step 1: Install Streaming Support

Install the SSE (Server-Sent Events) dependencies.

```python
from pathlib import Path
from lakelogic import DataProcessor

BASE = Path.cwd()
contract_path = BASE / "wikimedia_contract.yaml"
if not contract_path.exists():
    candidate = BASE / "examples" / "03_data_sources" / "streaming" / "sse_wikimedia" / "wikimedia_contract.yaml"
    if candidate.exists():
        BASE = candidate.parent
        contract_path = candidate

RUN_LIVE = False  # Set True when your streaming infrastructure is available

sample_events = [{'id': 1, 'type': 'edit', 'title': 'LakeLogic', 'user': 'alice', 'bot': False, 'timestamp': 1739580000}, {'id': 2, 'type': 'new', 'title': 'Data Quality', 'user': 'bob', 'bot': False, 'timestamp': 1739580060}]

if RUN_LIVE:
    print("Live streaming is disabled by default in this demo.")
    print("Set RUN_LIVE = True and configure credentials/endpoints to stream.")
else:
    processor = DataProcessor(contract=contract_path)
    result = processor.run(sample_events, source_path="sample")
    print(result)
    print(f"Good: {len(result.good)} | Bad: {len(result.bad)}")
```

##  Step 2: Use Your Contract

We have a `wikimedia_contract.yaml` file in this directory.
