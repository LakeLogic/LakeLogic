## Step 1: Install Dependencies

Install the required extras before running this example.


```python
from pathlib import Path
from lakelogic import DataProcessor

BASE = Path.cwd()
contract_path = BASE / "postgres_contract.yaml"
if not contract_path.exists():
    candidate = BASE / "examples" / "03_data_sources" / "databases" / "postgres" / "postgres_contract.yaml"
    if candidate.exists():
        BASE = candidate.parent
        contract_path = candidate

sample_rows = [
    {"order_id": 1001, "customer_id": 10, "order_date": "2026-02-01", "total_amount": 120.50},
    {"order_id": 1002, "customer_id": 11, "order_date": "2026-02-02", "total_amount": 0},
]

processor = DataProcessor(contract=contract_path)
result = processor.run(sample_rows, source_path="sample")
print(result)
print("Sample run complete. Configure your Postgres connection to run live.")
```

## Step 2: Review the Contract

Open `postgres_contract.yaml` to see the schema, source config, and quality rules.


## Step 3: Run the Contract

Run LakeLogic using the YAML contract. Update credentials as needed.
