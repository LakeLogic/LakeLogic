## Goals

## Step 1: Install Dependencies

Install the required extras before running this example.


```python
from pathlib import Path
from lakelogic import DataProcessor

BASE = Path.cwd()
contract_path = BASE / "sqlserver_contract.yaml"
if not contract_path.exists():
    candidate = BASE / "examples" / "03_data_sources" / "databases" / "sqlserver" / "sqlserver_contract.yaml"
    if candidate.exists():
        BASE = candidate.parent
        contract_path = candidate

sample_rows = [
    {"order_id": 2001, "customer_id": 20, "order_date": "2026-02-03", "total_amount": 88.00},
    {"order_id": None, "customer_id": 21, "order_date": "2026-02-04", "total_amount": 45.75},
]

processor = DataProcessor(contract=contract_path)
result = processor.run(sample_rows, source_path="sample")
print(result)
print("Sample run complete. Configure your SQL Server connection to run live.")
```

## Step 2: Review the Contract

Open `sqlserver_contract.yaml` to see the schema, source config, and quality rules.
