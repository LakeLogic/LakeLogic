## Step 1: Review contract.yaml

Open contract.yaml for the full production-ready template.


```python
from pathlib import Path
from lakelogic import DataProcessor


def resolve_example_dir(name: str) -> Path:
    cwd = Path.cwd()
    for base in [cwd] + list(cwd.parents):
        candidates = [
            base / name,
            base / "examples" / "07_production" / "ci_cd" / name,
            base / "lakelogic" / "examples" / "07_production" / "ci_cd" / name,
        ]
        for candidate in candidates:
            if candidate.exists():
                return candidate
    return cwd / name


base_path = resolve_example_dir("contract_template")
contract_path = base_path / "contract.yaml"

processor = DataProcessor(contract=contract_path)
# processor.run(base_path / "data" / "customers.csv")
```
