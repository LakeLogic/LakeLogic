## 🔐 Configuration

Databricks connection details are resolved from environment variables used in the YAML contract.

```python
import os

# Set actual values to run
os.environ["DATABRICKS_HOST"] = os.getenv("DATABRICKS_HOST", "your-host.azuredatabricks.net")
os.environ["DATABRICKS_HTTP_PATH"] = os.getenv("DATABRICKS_HTTP_PATH", "/sql/1.0/endpoints/your-endpoint")
os.environ["DATABRICKS_TOKEN"] = os.getenv("DATABRICKS_TOKEN", "dapi...")
```
