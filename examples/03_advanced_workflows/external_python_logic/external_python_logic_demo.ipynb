## Step 2: Run the Contract

Choose one contract and run it using the LakeLogic driver.


```python
import csv
from pathlib import Path

from lakelogic import DataProcessor

def resolve_example_dir(name: str) -> Path:
    cwd = Path.cwd()
    for base in [cwd] + list(cwd.parents):
        if base.name == name and base.exists():
            return base
        candidate = base / name
        if candidate.exists():
            return candidate
        candidate = base / "examples" / "06_advanced_workflows" / name
        if candidate.exists():
            return candidate
        candidate = base / "lakelogic" / "examples" / "06_advanced_workflows" / name
        if candidate.exists():
            return candidate
    return cwd / name

BASE = resolve_example_dir("external_python_logic")
contract_path = BASE / "contract_notebook.yaml"

data_path = BASE / "data" / "sales.csv"
rows = []
with data_path.open("r", newline="", encoding="utf-8") as handle:
    reader = csv.DictReader(handle)
    rows = list(reader)

processor = DataProcessor(contract=contract_path)
result = processor.run(rows, source_path=str(data_path))

def preview_frame(df, title=None, limit=5):
    if title:
        print(title)
    if df is None:
        print("<empty>")
        return
    if hasattr(df, "head"):
        try:
            print(df.head(limit))
            return
        except Exception:
            pass
    if hasattr(df, "limit"):
        try:
            print(df.limit(limit))
            return
        except Exception:
            pass
    print(df)

preview_frame(getattr(result, "good", None), title="Gold output (sample):")
```
