# LakeGuard: Getting Started Tutorial 🛡️

This notebook demonstrates the core capabilities of LakeGuard:
1. **Pre-Processing**: Deduplicating your data before validation.
2. **Schema Enforcement**: Validating data types and simple rules (e.g., email format).
3. **Quality Rules**: Enforcing business logic (e.g., age must be positive).
4. **Quarantine**: Isolating bad records while letting good data flow.
5. **Post-Processing**: Enriching your data with derived fields.


## 1. Setup
First, we'll install LakeGuard and import the necessary libraries.

Note: We no longer need to import `polars` or `pandas` directly for data loading; LakeGuard can handle it using the engine specified in your settings or code.

```python
# %pip install lakeguard[all]
import os
from lakeguard import DataProcessor

# You can set the engine via environment variable for global consistency
os.environ["LAKEGUARD_ENGINE"] = "duckdb"
```

## 2. Execute directly from Source
We have a CSV with some 'dirty' data. Instead of loading it manually, we call `run_source()` which uses the engine's optimized native readers.

```python
processor = DataProcessor(contract="contract.yaml")
df_good, df_bad = processor.run_source("data/sample_customers.csv")
```

## 3. Inspect the Results
Let's see what made it to our 'Silver' layer and what was sent to 'Quarantine'.

```python
print("✅ GOOD DATA (Silver):")
print(df_good)

print("\n🛑 BAD DATA (Quarantine):")
print(df_bad)
```

## 4. Analysis
Notice the following:
- **Deduplication**: One of the 'Alice' records was dropped (the older one by signup_date).
- **Schema Failure**: 'Charlie' was quarantined due to an invalid email format.
- **Rule Failure**: 'David' was quarantined because his age was -5 and status was 'inactive'.
- **Enrichment**: The `is_premium` column was correctly derived for the good data.
