## Business Scenario

Customer attributes change over time. Reporting and compliance require you to know what was true on a specific date, not just the latest snapshot.

## Value Proposition

- Preserve full history of attribute changes
- Track active vs. expired records
- Support time-travel analytics

---

## Goals

1. Load multiple snapshots
2. Close out old rows
3. Produce an SCD2 table


```python
import os
import shutil
import pandas as pd
from lakelogic import DataProcessor

os.environ["LAKELOGIC_ENGINE"] = "polars"
shutil.rmtree("output", ignore_errors=True)
```

```python
processor = DataProcessor(contract="contract.yaml")
snapshot1, _ = processor.run_source("data/dim_customers_snapshot1.csv")
processor.materialize(snapshot1)
pd.read_csv("output/gold_dim_customers/data.csv")
```

```python
snapshot2, _ = processor.run_source("data/dim_customers_snapshot2.csv")
processor.materialize(snapshot2)
pd.read_csv("output/gold_dim_customers/data.csv")
```
