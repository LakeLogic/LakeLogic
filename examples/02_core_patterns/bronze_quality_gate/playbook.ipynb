## Business Scenario

Raw signups and web leads arrive with missing emails, invalid ages, and inconsistent formats. Before this data hits Silver, you need a fast quality gate to stop obvious issues at ingestion.

## Value Proposition

- Stop bad data early without breaking the pipeline
- Quarantine invalid rows with clear reasons
- Keep Bronze ingestion lightweight and repeatable

---

## Goals

1. Load raw signups
2. Validate schema and row rules
3. Separate good and quarantined rows


```python
import os
from lakelogic import DataProcessor

os.environ["LAKELOGIC_ENGINE"] = "polars"
```

```python
processor = DataProcessor(contract="contract.yaml")
good_df, bad_df = processor.run_source("data/raw_signups.csv")
processor.last_report
```
