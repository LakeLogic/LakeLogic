# Late Arriving Reprocess
Partition output by date and overwrite only the affected partitions when late data arrives.

```python
import os
import shutil
from pathlib import Path
from lakeguard import DataProcessor

os.environ["LAKEGUARD_ENGINE"] = "polars"
shutil.rmtree("output", ignore_errors=True)
```

```python
processor = DataProcessor(contract="contract.yaml")
good_df, bad_df = processor.run_source("data/events_run1.csv")
processor.materialize(good_df, bad_df)
sorted(Path("output/silver_pos_sales_events").rglob("*.csv"))
```

```python
good_df, bad_df = processor.run_source("data/events_run2.csv")
processor.materialize(good_df, bad_df)
sorted(Path("output/silver_pos_sales_events").rglob("*.csv"))
```
