# LakeGuard Driver Notebook

This notebook runs the registry-driven pipeline driver against a registry.
It includes observability outputs (summary tables, metrics, and Prometheus).
Update the registry paths and window parameters as needed.


```python
# Run the driver (edit layers/window as needed)
import subprocess
from pathlib import Path

OUTPUT_DIR = Path('examples/insurance_elt/output/run_logs')
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

cmd = [
    'lakeguard-driver',
    '--registry', 'examples/insurance_elt/contracts/insurance/_registry.yaml',
    '--reference-registry', 'examples/insurance_elt/contracts/shared/reference/_registry.yaml',
    '--gold-registry', 'examples/insurance_elt/contracts/insurance/warehouse/_registry.yaml',
    '--layers', 'reference,bronze,silver,gold',
    '--window', 'last_success',
    '--strict-layer-order',
    '--summary-table', 'lakeguard.pipeline_runs',
    '--summary-backend', 'duckdb',
    '--summary-database', str(OUTPUT_DIR / 'lakeguard_pipeline_runs.duckdb'),
    '--metrics-path', str(OUTPUT_DIR / 'pipeline_metrics.json'),
]
print(' '.join(cmd))
subprocess.run(cmd, check=True)
```
