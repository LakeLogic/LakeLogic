# 🌐 REST API Data Extraction

This notebook demonstrates how to extract data from a JSON REST API using a LakeLogic YAML contract file.

---

## 🎯 Goals

1. Connect to an external JSON API
2. Execute extraction using `weather_contract.yaml` directly
3. Fetch real-time weather stats

## 🚀 Setup

## ▶️ Run the Extraction

Pass the YAML contract path directly to the `DataProcessor`.

```python
from lakelogic.core.processor import DataProcessor

print("🚀 Fetching weather data using weather_contract.yaml...")

processor = DataProcessor(contract="weather_contract.yaml")
result = processor.run()

print("\n✅ Weather data extracted successfully to ./data/bronze/weather/")
```

## 🎉 Verify Results

LakeLogic stores data in Parquet format by default.

```python
import polars as pl

df = pl.read_parquet("./data/bronze/weather/*.parquet")
print(df.head())
```
