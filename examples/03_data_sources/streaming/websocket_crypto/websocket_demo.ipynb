## Business Scenario

Crypto price feeds and market data are delivered over WebSockets. You need schema validation before storing ticks.

## Value Proposition

- Real-time WebSocket ingestion
- Contract-based schema validation
- Clean downstream time-series data

---

## Goals

1. Connect to a WebSocket feed
2. Validate ticks
3. Persist clean events


## Step 1: Install Dependencies

Install the required extras before running this example.


```python
from pathlib import Path
from lakelogic import DataProcessor

BASE = Path.cwd()
contract_path = BASE / "websocket_contract.yaml"
if not contract_path.exists():
    candidate = BASE / "examples" / "03_data_sources" / "streaming" / "websocket_crypto" / "websocket_contract.yaml"
    if candidate.exists():
        BASE = candidate.parent
        contract_path = candidate

RUN_LIVE = False  # Set True when your streaming infrastructure is available

sample_events = [{'type': 'ticker', 'product_id': 'BTC-USD', 'price': 42000.12, 'time': '2026-02-15T10:50:00'}, {'type': 'ticker', 'product_id': 'BTC-USD', 'price': 42010.55, 'time': '2026-02-15T10:51:00'}]

if RUN_LIVE:
    print("Live streaming is disabled by default in this demo.")
    print("Set RUN_LIVE = True and configure credentials/endpoints to stream.")
else:
    processor = DataProcessor(contract=contract_path)
    result = processor.run(sample_events, source_path="sample")
    print(result)
    print(f"Good: {len(result.good)} | Bad: {len(result.bad)}")
```

## Step 2: Review the Contract

Open `websocket_contract.yaml` to see the schema, source config, and quality rules.

Run LakeLogic using the YAML contract. Update credentials as needed.
