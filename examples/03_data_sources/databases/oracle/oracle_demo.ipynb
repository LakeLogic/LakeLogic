## Step 1: Install Dependencies

Install the required extras before running this example.


```python
from pathlib import Path
from lakelogic import DataProcessor

BASE = Path.cwd()
contract_path = BASE / "oracle_contract.yaml"
if not contract_path.exists():
    candidate = BASE / "examples" / "03_data_sources" / "databases" / "oracle" / "oracle_contract.yaml"
    if candidate.exists():
        BASE = candidate.parent
        contract_path = candidate

sample_rows = [
    {"ORDER_ID": 3001, "CUSTOMER_ID": 30, "ORDER_DATE": "2026-02-05", "TOTAL_AMOUNT": 210.00},
    {"ORDER_ID": None, "CUSTOMER_ID": 31, "ORDER_DATE": "2026-02-06", "TOTAL_AMOUNT": 75.00},
]

processor = DataProcessor(contract=contract_path)
result = processor.run(sample_rows, source_path="sample")
print(result)
print("Sample run complete. Configure your Oracle connection to run live.")
```
