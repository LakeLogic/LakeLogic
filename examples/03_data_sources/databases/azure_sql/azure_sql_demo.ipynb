## Step 1: Install Dependencies

Install the required extras before running this example.


```python
from pathlib import Path
from lakelogic import DataProcessor

BASE = Path.cwd()
contract_path = BASE / "azure_sql_contract.yaml"
if not contract_path.exists():
    candidate = BASE / "examples" / "03_data_sources" / "databases" / "azure_sql" / "azure_sql_contract.yaml"
    if candidate.exists():
        BASE = candidate.parent
        contract_path = candidate

sample_rows = [
    {"order_id": 4001, "customer_id": 40, "order_date": "2026-02-07", "total_amount": 55.00},
    {"order_id": 4002, "customer_id": 41, "order_date": "2026-02-08", "total_amount": -5.00},
]

processor = DataProcessor(contract=contract_path)
result = processor.run(sample_rows, source_path="sample")
print(result)
print("Sample run complete. Configure your Azure SQL connection to run live.")
```

## Step 2: Review the Contract

Open `azure_sql_contract.yaml` to see the schema, source config, and quality rules.

Run LakeLogic using the YAML contract. Update credentials as needed.
