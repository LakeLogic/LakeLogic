## Step 1: Install Dependencies

Install the required extras before running this example.


```python
from pathlib import Path
from lakelogic import DataProcessor

BASE = Path.cwd()
contract_path = BASE / "azuresql_cdc_contract.yaml"
if not contract_path.exists():
    candidate = BASE / "examples" / "03_data_sources" / "databases" / "azure_sql" / "cdc" / "azuresql_cdc_contract.yaml"
    if candidate.exists():
        BASE = candidate.parent
        contract_path = candidate

sample_rows = [
    {
        "order_id": 5001,
        "customer_id": 50,
        "order_date": "2026-02-09",
        "total_amount": 99.99,
        "status": "confirmed",
        "payment_method": "credit_card",
        "shipping_address": "123 Main St, Seattle",
        "created_at": "2026-02-09T08:00:00",
        "updated_at": "2026-02-09T09:00:00",
    },
    {
        "order_id": 5002,
        "customer_id": 51,
        "order_date": "2026-02-10",
        "total_amount": 10.00,
        "status": "pending",
        "payment_method": "paypal",
        "shipping_address": "55 Pine St, Portland",
        "created_at": "2026-02-10T08:00:00",
        "updated_at": "2026-02-10T08:30:00",
    },
]

processor = DataProcessor(contract=contract_path)
result = processor.run(sample_rows, source_path="sample")
print(result)
print("Sample run complete. Configure your Azure SQL CDC connection to run live.")
```
