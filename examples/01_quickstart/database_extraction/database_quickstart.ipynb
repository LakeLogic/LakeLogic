## 🏢 Business Scenario

Most organizations have critical data locked in traditional **SQL Databases** (ERP, CRM, Legacy internal apps). To unlock advanced analytics, machine learning, and high-performance reporting, this data needs to be moved into a **Modern Data Lakehouse** (like Delta Lake or Iceberg).

Common challenges include:
*   **Schema Drift**: The database table structure changes, breaking your ETL scripts.
*   **Data Quality**: Null emails or corrupt IDs from the source system polluting your lakehouse.
*   **Maintenance**: Managing dozens of separate extraction scripts for different tables.

## 💡 Value Proposition

*   **Contract-Driven Extraction**: You define the "Desired State" of your data in a YAML file. LakeLogic handles the connection, extraction, and validation automatically.
*   **Unified Engine**: Use the same contract to extract from local SQLite, Azure SQL, Postgres, or Oracle. The logic stays the same; only the connection details change.
*   **Built-in Delta Lake Storage**: LakeLogic automatically formats your data into optimized Delta tables, ready for Spark, DuckDB, or BI tools.
*   **Low Code**: Go from zero to a production-ready ingestion pipeline in minutes without writing complex Python orchestration logic.

## 🚀 Step 1: Setup Local Database

For this tutorial, we will use **SQLite** (included with Python). Let's create some sample data.

```python
import sqlite3
import os

db_file = 'example.db'
conn = sqlite3.connect(db_file)
c = conn.cursor()
c.execute('CREATE TABLE IF NOT EXISTS users (id INTEGER, name TEXT, email TEXT)')
c.execute('DELETE FROM users')
c.execute('INSERT INTO users VALUES (1, "Alice", "alice@example.com")')
c.execute('INSERT INTO users VALUES (2, "Bob", "bob@example.com")')
c.execute('INSERT INTO users VALUES (3, "Charlie", "charlie@example.com")')
conn.commit()
conn.close()
print("✅ Local database setup complete!")
```

## 📝 Step 2: Use Your Contract

We have a `users_contract.yaml` file in this directory. LakeLogic can read this file directly.

```python
with open('users_contract.yaml', 'r') as f:
    print("📄 users_contract.yaml content:")
    print("-----------------------------")
    print(f.read())
```

## ▶️ Step 3: Run the Extraction

Instead of loading JSON/dict, we just pass the file path to `DataProcessor`.

```python
from lakelogic.core.processor import DataProcessor

# Run LakeLogic by passing the YAML path directly
processor = DataProcessor(contract="users_contract.yaml")
result = processor.run()

print("✅ Data extracted successfully using YAML contract path!")
```

## 🎉 Check Your Results

Read the resulting Delta table using `polars`.

```python
import polars as pl

df = pl.read_delta("./data/bronze/users/")
print(df)
```
