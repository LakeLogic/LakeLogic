## Business Scenario

CSV drops are common for partner feeds and exports. You need schema validation before loading them.

## Value Proposition

- Define CSV schema in YAML
- Validate rows on ingestion
- Produce clean outputs

---

## Goals

1. Load a CSV file
2. Validate against contract
3. Inspect results


## Step 2: Review the Contract

Open `csv_contract.yaml` to see the schema, source config, and quality rules.


## Step 3: Run the Contract

Run LakeLogic using the YAML contract. Update credentials as needed.


```python
from pathlib import Path
import csv
from lakelogic import DataProcessor

BASE = Path.cwd()
contract_path = BASE / "csv_contract.yaml"
if not contract_path.exists():
    candidate = BASE / "examples" / "03_data_sources" / "files" / "csv" / "csv_contract.yaml"
    if candidate.exists():
        BASE = candidate.parent
        contract_path = candidate

data_dir = BASE / "data"
data_dir.mkdir(parents=True, exist_ok=True)

csv_path = data_dir / "sample.csv"
if not csv_path.exists():
    rows = [
        {"id": 1, "email": "alice@example.com"},
        {"id": 2, "email": ""},
        {"id": 3, "email": "bob@example.com"},
    ]
    with open(csv_path, "w", newline="", encoding="utf-8") as handle:
        writer = csv.DictWriter(handle, fieldnames=["id", "email"])
        writer.writeheader()
        writer.writerows(rows)

processor = DataProcessor(contract=contract_path)
result = processor.run_source(csv_path)

print(result)
print(f"Good: {len(result.good)} | Bad: {len(result.bad)}")
```
