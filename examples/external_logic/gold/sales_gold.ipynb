# Sales Gold (External Notebook)
This notebook reads validated input from LakeGuard and writes a Gold output.


```python
import pandas as pd
params = LAKEGUARD_PARAMS
input_path = params.get('lakeguard_input_path')
output_path = params.get('lakeguard_output_path')

df = pd.read_csv(input_path)
agg = df.groupby('salesperson_id', as_index=False)['amount'].sum()
agg = agg.rename(columns={'amount': 'total_amount'})
agg['processed_at'] = pd.Timestamp.utcnow().tz_localize('UTC')

if output_path:
    agg.to_csv(output_path, index=False)

agg.head()
```
