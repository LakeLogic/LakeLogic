## 🌊 Step 3: Your First Stream

We use `DataProcessor` pointing directly to the YAML file.

```python
from lakelogic.core.processor import DataProcessor

# Pass the YAML path directly
processor = DataProcessor(contract="wikimedia_contract.yaml")

print("🌊 Streaming Wikipedia edits in real-time...")
print("Press 'Stop' in the notebook toolbar to end the stream")
print()

# Note: For streaming, we can use processor.stream() 
# or if the contract is set up for materialization, processor.run()
# Here we'll iterate manually to show the events

event_count = 0
try:
    # Access the connector through the adapter
    for event in processor.adapter.stream():
        event_count += 1
        
        title = event.get('title', 'N/A')
        user = event.get('user', 'N/A')
        event_type = event.get('type', 'unknown')
        
        print(f"[{event_count:04d}] {event_type:10s} | {title[:50]:50s} | by {user}")
        
        if event_count >= 20:
            print()
            print(f"✅ Processed {event_count} events!")
            break
finally:
    processor.adapter.close()
```
